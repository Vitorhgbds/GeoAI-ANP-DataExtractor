```python
import re
import pandas as pd

# Read the file
with open("../downloads/bacia de são luís/Categoria-2/2-FIST-1-MA/AGP/2FIST0001MA_agp.txt", "r", encoding="utf-8", errors="ignore") as f:
    content = f.read()

# Find the table section
match = re.search(r"RESUMO DAS ROCHAS ENCONTRADAS NO POCO -\s*\n(.*?)(?:\n\s*\n|\Z)", content, re.DOTALL)
if match:
    table_text = match.group(1)
    # Extract rows using regex
    rows = re.findall(r"^\s*(\d+)\s+([A-Z .]+?)\s*=\s*([\d.]+)\s*M\s*([\d.]+)\s*%", table_text, re.MULTILINE)
    # Create DataFrame
    df = pd.DataFrame(rows, columns=["COD", "ROCHA", "METROS", "PERCENTUAL"])
    # Cast columns to appropriate types
    df["COD"] = df["COD"].astype(int)
    df["METROS"] = df["METROS"].astype(float)
    df["PERCENTUAL"] = df["PERCENTUAL"].astype(float)
    print(df)
else:
    print("Table not found.")
```

```python
import re
import pandas as pd

with open(r"../downloads/bacia de alagoas/Categoria-1/1-CT-1-AL/AGP\1CT0001AL_agp.txt", "r", encoding="utf-8", errors="ignore") as f:
    content = f.read()

# Find the LITOLOGIA table section
match = re.search(r"LITOLOGIA -\s*\n(?:-+\s*\n)?\s*(.*?)(?=\n\s*\n|$)", content, re.DOTALL)
if match:
    table_text = match.group(1)
    # Improved regex: TOPO is optional, all columns after ROCHA are optional
    rows = re.findall(
        r"^\s*(?:(?P<topo>[\d.]+)\s*\([^)]+\))?\s*"
        r"(?P<base>[\d.]+)\s*\([^)]+\)\s*"
        r"(?P<cod>\d+)\s+(?P<rocha>[A-Z]+)"
        r"(?:\s+(?P<cor>[A-Z]+))?"
        r"(?:\s+(?P<tonalidade>[A-Z]+))?"
        r"(?:\s+(?P<granulometria>[A-Z]+))?"
        r"(?:\s+(?P<arredondamento>[A-Z]+))?"
        r"\s*$",
        table_text,
        re.MULTILINE
    )
    data = []
    for row in rows:
        topo, base, cod, rocha, cor, tonalidade, granulometria, arredondamento = row
        data.append({
            "TOPO": float(topo) if topo else None,
            "BASE": float(base) if base else None,
            "COD": cod,
            "ROCHA": rocha,
            "COR": cor,
            "TONALIDADE": tonalidade,
            "GRANULOMETRIA": granulometria,
            "ARREDONDAMENTO": arredondamento
        })
    df = pd.DataFrame(data)
    print(df)
else:
    print("LITOLOGIA table not found.")
    
```

```python
import re
import pandas as pd

# Read the file
with open("../downloads/bacia de são luís/Categoria-2/2-FIST-1-MA/AGP/2FIST0001MA_agp.txt", "r", encoding="utf-8", errors="ignore") as f:
    content = f.read()

# Find the table section
match = re.search(
            r"POCO\s*:\s*(?P<POCO>.+?)\s*\n.*?BACIA\s*:\s*(?P<BACIA>.+?)\s*\(",
            content,
            re.DOTALL
        )
well = match.group("POCO").strip().split("\n")[0].strip() if match else "N/A"
basin = match.group("BACIA").strip().split("\n")[0].strip() if match else "N/A"

print(f"Well: {well}\n-----------------\n Basin: {basin}")
```
