```python
import re
import pandas as pd

# Read the file
with open("../downloads/bacia de são luís/Categoria-2/2-FIST-1-MA/AGP/2FIST0001MA_agp.txt/2FIST0001MA_agp.txt", "r", encoding="utf-8", errors="ignore") as f:
    content = f.read()

# Find the table section
match = re.search(r"RESUMO DAS ROCHAS ENCONTRADAS NO POCO -\s*\n(.*?)(?:\n\s*\n|\Z)", content, re.DOTALL)
if match:
    table_text = match.group(1)
    # Extract rows using regex
    rows = re.findall(r"^\s*(\d+)\s+([A-Z .]+?)\s*=\s*([\d.]+)\s*M\s*([\d.]+)\s*%", table_text, re.MULTILINE)
    # Create DataFrame
    df = pd.DataFrame(rows, columns=["COD", "ROCHA", "METROS", "PERCENTUAL"])
    # Cast columns to appropriate types
    df["COD"] = df["COD"].astype(int)
    df["METROS"] = df["METROS"].astype(float)
    df["PERCENTUAL"] = df["PERCENTUAL"].astype(float)
    print(df)
else:
    print("Table not found.")
```
